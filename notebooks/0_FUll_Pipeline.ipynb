# EDA – EXPLORATORY DATA ANALYSIS

```python
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(exclude=np.number).columns

print("Numerical Columns:")
print(numerical_columns.tolist())

print("\nCategorical Columns:")
print(categorical_columns.tolist())
```

```python
#Distribution of numerical variables
df[numerical_columns].hist(
    figsize=(15, 10),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
#Boxplots of numerical variables
for column in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f'Boxplot of {column}')
    plt.show()
```

```python
# Select only meaningful categorical features (excluding IDs, dates, and long text)
categorical_columns = [
    'Project_Name',
    'Employee_Department',
    'Employee_Job_Position',
    'Task_Priority',
    'Task_Stage'
]

# Display value counts for each categorical feature
for column in categorical_columns:
    print(f"\n========== {column} ==========")
    display(df[column].value_counts())
```

```python
import matplotlib.pyplot as plt

categorical_columns = [
    'Project_Name',
    'Employee_Department',
    'Employee_Job_Position',
    'Task_Priority',
    'Task_Stage'
]

for column in categorical_columns:
    plt.figure(figsize=(8, 5))

    # Plot top 10 categories
    top_counts = df[column].value_counts().head(10)
    top_counts.plot(kind='bar', color='skyblue', edgecolor='black')

    plt.title(f'Top 10 Distribution of {column}', fontsize=12, fontweight='bold')
    plt.xlabel(column, fontsize=10)
    plt.ylabel('Count', fontsize=10)

    # Ensure labels fit cleanly without overlapping
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

# CORRELATION ANALYSIS

```python
correlation_matrix = df[numerical_columns].corr()

display(correlation_matrix)
```

```python
import seaborn as sns
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```
